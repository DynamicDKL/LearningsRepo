Naming conventions for variables

```python
# A variable name must start with a letter or the underscore character
# A variable name cannot start with a number
# A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
# Variable names are case-sensitive (age, Age and AGE are three different variables)
# A variable name cannot be any of the Python keywords.
# no white  spaces in between
# variables are changeable.
# Have habit of giving proper meaningful names to the variables always
```

String Slicing is important for the string data type.

string slicing  b[0:4], 0 is included and 4 is excluded 0,1,2 and 3 are output.

```python
b[0:4]
```

Slicing  with jump

```python
b[0:4:2]
```

Negative Indexing 

```python
# first character starts with -(len), and increments till the last character. for 'python' -6 to -1
```

```python
b[0:4], b[-6:-2] # default jump is +1
```

reverse a string; 

```python
b[::-1] # default wil be 0 for start and len(a) to end
```

```python
#first 3 chars in reverse order
b[-4::-1] # here moving back the end will the fisrt index which is -6
```
